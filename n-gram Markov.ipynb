```python
import random
import math
from collections import defaultdict, Counter

# Texte d'entraînement
text = "This is Time to go to Home"
n = 3  # trigrammes

# Création du modèle de Markov avec compte des transitions
markov_model = defaultdict(Counter)

# Apprentissage : construction des trigrammes
for i in range(len(text) - n):
    prefix = text[i:i+n]
    next_char = text[i+n]
    markov_model[prefix][next_char] += 1

# Fonction de génération de texte
def generate_text(model, start_seq, length=50):
    generated = start_seq
    current = start_seq[-n:]
    for _ in range(length):
        next_chars = model.get(current)
        if not next_chars:
            break
        next_char = random.choices(
            population=list(next_chars.keys()),
            weights=list(next_chars.values())
        )[0]
        generated += next_char
        current = generated[-n:]
    return generated

# Fonction pour calculer la perplexité
def calculate_perplexity(model, text, n):
    N = 0
    log_prob = 0.0
    for i in range(len(text) - n):
        prefix = text[i:i+n]
        next_char = text[i+n]
        next_chars = model.get(prefix)
        if not next_chars or sum(next_chars.values()) == 0:
            continue
        total = sum(next_chars.values())
        prob = next_chars[next_char] / total if next_char in next_chars else 1e-6
        log_prob += math.log(prob)
        N += 1
    perplexity = math.exp(-log_prob / N) if N > 0 else float('inf')
    return perplexity

# Texte de départ
start_seq = "This is T"
generated_text = generate_text(markov_model, start_seq)

# Affichage du texte généré
print("Generated Text:")
print(generated_text)

# Calcul des métriques
perplexity = calculate_perplexity(markov_model, text, n)
print("\n=== Metrics ===")
print(f"Perplexity on training data: {perplexity:.4f}")
```
